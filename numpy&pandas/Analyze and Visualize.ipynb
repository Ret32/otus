# I. Numpy

### Импортируйте NumPy

```python
import numpy as np
```

### Создайте одномерный массив размера 10, заполненный нулями и пятым элемент равным 1. Трансформируйте в двумерный массив.

```python
arr = np.zeros(10, dtype=int)
arr[4] = 1
arr_2d = arr.reshape(2, 5)

print(arr_2d)
```

### Создайте одномерный массив со значениями от 10 до 49 и разверните его (первый элемент становится последним). Найдите в нем все четные элементы.

```python
arr = np.arange(10, 50)
arr_reversed = arr[::-1]
even_elements = arr_reversed[arr_reversed % 2 == 0]

print("Развернутый массив:", arr_reversed)
print("Четные элементы:", even_elements)
```

### Создайте двумерный массив 3x3 со значениями от 0 до 8

```python
matrix_3x3 = np.arange(9).reshape(3, 3)
print(matrix_3x3)
```

### Создайте массив 4x3x2 со случайными значениями. Найти его минимум и максимум.

```python
random_array = np.random.rand(4, 3, 2)
min_value = random_array.min()
max_value = random_array.max()

print("Случайный массив 4x3x2:\n", random_array)
print("Минимальное значение:", min_value)
print("Максимальное значение:", max_value)
```

### Создайте два двумерных массива размерами 6x4 и 4x3 и произведите их матричное умножение. 

```python
matrix_6x4 = np.random.randint(0, 10, (6, 4))
matrix_4x3 = np.random.randint(0, 10, (4, 3))
result_matrix = np.dot(matrix_6x4, matrix_4x3)

print("Матрица 6x4:\n", matrix_6x4)
print("Матрица 4x3:\n", matrix_4x3)
print("Результат умножения (6x3):\n", result_matrix)
```

### Создайте случайный двумерный массив 7x7, найти у него среднее и стандартное оклонение. Нормализуйте этот массив.

```python
matrix_7x7 = np.random.rand(7, 7)
mean_value = matrix_7x7.mean()
std_deviation = matrix_7x7.std()
normalized_matrix = (matrix_7x7 - mean_value) / std_deviation

print("Исходный массив 7x7:\n", matrix_7x7)
print("Среднее значение:", mean_value)
print("Стандартное отклонение:", std_deviation)
print("Нормализованный массив:\n", normalized_matrix)
```

# II. Pandas

### Импортируйте: pandas, matplotlib, seaborn

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Загрузите датасет Tips из набора датасетов seaborn

```python
tips = sns.load_dataset("tips")
```

### Посмотрите на первые 5 строчек

```python
print(tips.head())
```

### Узнайте сколько всего строчек и колонок в данных

```python
rows, cols = tips.shape

print(f"Количество строк: {rows}")
print(f"Количество колонок: {cols}")
```

### Проверьте есть ли пропуски в данных

```python
is_null = tips.isnull().sum()
print ("Пропущенные данные:\n", is_null)
```

### Посмотрите на распределение числовых признаков

```python
numeric_columns = tips.select_dtypes(include=['float64', 'int64']).columns
print("Числовые признаки:", list(numeric_columns))

for col in numeric_columns:
    plt.figure(figsize=(8, 4))
    sns.histplot(tips[col], kde=True)
    plt.title(f'Распределение числового признака: {col}')
    plt.xlabel(col)
    plt.ylabel('Частота')
    plt.show()
```

### Найдите максимальное значение 'total_bill'

```python
max_total_bill = tips['total_bill'].max()
print("Максимальное значение 'total_bill':", max_total_bill)
```

### Найдите количество курящих людей

```python
smoker_count = tips[tips['smoker'] == 'Yes'].shape[0]
print("Количество курящих людей:", smoker_count)
```

### Узнайте какой средний 'total_bill' в зависимости от 'day'

```python
mean_total_bill_by_day = tips.groupby('day', observed=False)['total_bill'].mean()
print("Средний 'total_bill' по дням:", mean_total_bill_by_day)
```

### Отберите строчки с 'total_bill' больше медианы и узнайте какой средний 'tip' в зависимости от 'sex'

```python
median_total_bill = tips['total_bill'].median()

filtered_tips = tips[tips['total_bill'] > median_total_bill]

mean_tip_by_sex = filtered_tips.groupby('sex', observed=False)['tip'].mean()

print("Средний 'tip' по полу для строк, где 'total_bill' больше медианы:\n", mean_tip_by_sex)
```

### Преобразуйте признак 'smoker' в бинарный (0-No, 1-Yes)

```python
tips['smoker_bin'] = tips['smoker'].map({'No': 0, 'Yes': 1})

print(tips[['smoker', 'smoker_bin']].head())
```

# III. Visualization

### Постройте гистограмму распределения признака 'total_bill'

```python
plt.figure(figsize=(10, 6))
sns.histplot(tips['total_bill'], kde=True)
plt.title('Гистограмма распределения total_bill')
plt.xlabel('Total Bill')
plt.ylabel('Частота')
plt.show()
```

### Постройте scatterplot, представляющий взаимосвязь между признаками 'total_bill' и 'tip'

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=tips, x='total_bill', y='tip')
plt.title('Scatterplot: взаимосвязь total_bill и tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

### Постройте pairplot

```python
sns.pairplot(tips)
plt.show()
```

### Постройте график взаимосвязи между признаками 'total_bill' и 'day'

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=tips, x='total_bill', y='day')
plt.title('Взаимосвязь между total_bill и day')
plt.xlabel('Total Bill')
plt.ylabel('Day')
plt.show()
```

### Постройте две гистограммы распределения признака 'tip' в зависимости от категорий 'time'

```python
plt.figure(figsize=(12, 6))

# Гистограмма для обеденного времени
plt.subplot(1, 2, 1)
sns.histplot(data=tips[tips['time'] == 'Lunch'], x='tip', kde=True, color='skyblue')
plt.title('Распределение признака tip во время обеда')
plt.xlabel('Tip')
plt.ylabel('Частота')

# Гистограмма для ужина
plt.subplot(1, 2, 2)
sns.histplot(data=tips[tips['time'] == 'Dinner'], x='tip', kde=True, color='salmon')
plt.title('Распределение признака tip во время ужина')
plt.xlabel('Tip')
plt.ylabel('Частота')

plt.tight_layout()
plt.show()
```

### Постройте два графика scatterplot, представляющих взаимосвязь между признаками 'total_bill' и 'tip' один для Male, другой для Female и раскрасьте точки в зависимоти от признака 'smoker'

```python
male_data = tips[tips['sex'] == 'Male']
female_data = tips[tips['sex'] == 'Female']

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

sns.scatterplot(data=male_data, x='total_bill', y='tip', hue='smoker', palette='coolwarm', ax=axes[0])
axes[0].set_title('Мужчины')
axes[0].set_xlabel('Total Bill')
axes[0].set_ylabel('Tip')

sns.scatterplot(data=female_data, x='total_bill', y='tip', hue='smoker', palette='coolwarm', ax=axes[1])
axes[1].set_title('Женщины')
axes[1].set_xlabel('Total Bill')
axes[1].set_ylabel('Tip')

plt.tight_layout()
plt.show()
```
